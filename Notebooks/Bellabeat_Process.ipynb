## Step 2: Data Quality Inspection

```python
# Step 2A: Import daily activity datasets

# Identify the two dailyActivity CSV files
daily_files = [
    file for file in selected_files
    if file.stem == "dailyActivity_merged"
]

# Identify each time period using its folder name
period1_path = next(
    file for file in daily_files
    if file.parent.name == "Fitabase Data 3.12.16-4.11.16"
)

period2_path = next(
    file for file in daily_files
    if file.parent.name == "Fitabase Data 4.12.16-5.12.16"
)

# Load the original CSV files into pandas DataFrames
daily_1 = pd.read_csv(period1_path)
daily_2 = pd.read_csv(period2_path)

# Display dataset dimensions
print("Period 1:", daily_1.shape)
print("Period 2:", daily_2.shape)

# Display the first five records from each dataset
print("\nPeriod 1 Preview:")
display(daily_1.head())

print("\nPeriod 2 Preview:")
display(daily_2.head())
```

```python
# Step 2B: Inspect column names and data types

# Display column names for each dataset
print("Period 1 Columns:")
print(daily_1.columns.tolist())

print("\nPeriod 2 Columns:")
print(daily_2.columns.tolist())

# Check whether column names match exactly
print("\nColumn names match:",
      daily_1.columns.equals(daily_2.columns))

# Display data types and non-null counts
print("\nPeriod 1 Information:")
daily_1.info()

print("\nPeriod 2 Information:")
daily_2.info()
```

```python
# Step 2C: Check for missing values

# Calculate missing values for Period 1
missing_1 = daily_1.isnull().sum()

# Calculate missing values for Period 2
missing_2 = daily_2.isnull().sum()

# Calculate missing percentages
missing_pct_1 = (missing_1 / len(daily_1)) * 100
missing_pct_2 = (missing_2 / len(daily_2)) * 100

# Create a comparison table
missing_summary = pd.DataFrame({
    "Period 1 Missing": missing_1,
    "Period 1 Missing (%)": missing_pct_1,
    "Period 2 Missing": missing_2,
    "Period 2 Missing (%)": missing_pct_2
})

# Display results
display(missing_summary.round(2))

# Display total missing values
print("Total missing values:")
print("Period 1:", missing_1.sum())
print("Period 2:", missing_2.sum())
```

```python
# Step 2D: Check for duplicate records

# 1. Check for exact duplicate rows
exact_duplicates_1 = daily_1.duplicated().sum()
exact_duplicates_2 = daily_2.duplicated().sum()

print("Exact Duplicate Records:")
print("Period 1:", exact_duplicates_1)
print("Period 2:", exact_duplicates_2)

# 2. Check for duplicate participant-date combinations
key_columns = ["Id", "ActivityDate"]

duplicate_keys_1 = daily_1.duplicated(
    subset=key_columns
).sum()

duplicate_keys_2 = daily_2.duplicated(
    subset=key_columns
).sum()

print("\nDuplicate Participant-Date Records:")
print("Period 1:", duplicate_keys_1)
print("Period 2:", duplicate_keys_2)

# 3. Display repeated participant-date records
print("\nPeriod 1 Duplicate Details:")
display(
    daily_1[
        daily_1.duplicated(
            subset=key_columns, keep=False
        )
    ]
)

print("\nPeriod 2 Duplicate Details:")
display(
    daily_2[
        daily_2.duplicated(
            subset=key_columns, keep=False
        )
    ]
)
```

```python
# Step 2E: Inspect date ranges

# Create copies to preserve the original imported data
daily_check_1 = daily_1.copy()
daily_check_2 = daily_2.copy()

# Convert ActivityDate to datetime
for df in [daily_check_1, daily_check_2]:
    df["ActivityDate"] = pd.to_datetime(
        df["ActivityDate"],
        format="%m/%d/%Y",
        errors="coerce"
    )

# Check for dates that could not be converted
print("Invalid or missing dates:")
print("Period 1:", daily_check_1["ActivityDate"].isna().sum())
print("Period 2:", daily_check_2["ActivityDate"].isna().sum())

# Display date ranges
print("\nPeriod 1 Date Range:")
print("Earliest:", daily_check_1["ActivityDate"].min())
print("Latest:", daily_check_1["ActivityDate"].max())

print("\nPeriod 2 Date Range:")
print("Earliest:", daily_check_2["ActivityDate"].min())
print("Latest:", daily_check_2["ActivityDate"].max())

# Count unique participants and recorded dates
print("\nPeriod 1:")
print("Unique participants:", daily_check_1["Id"].nunique())
print("Unique dates:", daily_check_1["ActivityDate"].nunique())

print("\nPeriod 2:")
print("Unique participants:", daily_check_2["Id"].nunique())
print("Unique dates:", daily_check_2["ActivityDate"].nunique())
```

```python
# Check overlapping calendar dates
dates_1 = set(daily_check_1["ActivityDate"].dropna())
dates_2 = set(daily_check_2["ActivityDate"].dropna())

overlapping_dates = dates_1.intersection(dates_2)

print("Overlapping calendar dates:", len(overlapping_dates))
print("Dates:", sorted(overlapping_dates))

# Check overlapping participant-date records
keys = ["Id", "ActivityDate"]

overlapping_records = daily_check_1[keys].merge(
    daily_check_2[keys],
    on=keys,
    how="inner",
    validate="one_to_one"
)

print("\nOverlapping participant-date records:",
      len(overlapping_records))

display(overlapping_records.head(10))
```

```python
# Step 2E.3: Compare overlapping participant-date records

# Identify the columns containing activity measurements
measurement_columns = [
    col for col in daily_check_1.columns
    if col not in ["Id", "ActivityDate"]
]

# Join records from both periods using participant and date
overlap_comparison = daily_check_1.merge(
    daily_check_2,
    on=["Id", "ActivityDate"],
    how="inner",
    suffixes=("_p1", "_p2"),
    validate="one_to_one"
)

# Compare measurements across the two periods
# np.isclose accounts for minor floating-point differences
match_results = pd.DataFrame({
    col: np.isclose(
        overlap_comparison[f"{col}_p1"],
        overlap_comparison[f"{col}_p2"],
        equal_nan=True
    )
    for col in measurement_columns
})

# Identify records whose measurements all match
overlap_comparison["All_Measurements_Match"] = (
    match_results.all(axis=1)
)

# Display comparison results
print("Total overlapping records:", len(overlap_comparison))

print("Records with matching measurements:",
      overlap_comparison["All_Measurements_Match"].sum())

print("Records with different measurements:",
      (~overlap_comparison["All_Measurements_Match"]).sum())

# Show which measurement columns differ
print("\nNumber of differences by column:")
display((~match_results).sum().sort_values(ascending=False))

# Preview records with conflicting measurements
print("\nPreview of conflicting records:")
display(
    overlap_comparison.loc[
        ~overlap_comparison["All_Measurements_Match"],
        [
            "Id",
            "ActivityDate",
            "TotalSteps_p1",
            "TotalSteps_p2",
            "Calories_p1",
            "Calories_p2"
        ]
    ].head(10)
)
```

```python
# Step 2E.4: Evaluate overlapping records

# Select important activity measurements
compare_columns = [
    "TotalSteps",
    "TotalDistance",
    "Calories"
]

# Compare Period 1 and Period 2 values
comparison_results = []

for col in compare_columns:

    # Extract measurements from both periods
    p1 = overlap_comparison[f"{col}_p1"]
    p2 = overlap_comparison[f"{col}_p2"]

    # Count how the measurements compare
    comparison_results.append({
        "Measurement": col,
        "Period 2 Higher": (p2 > p1).sum(),
        "Period 1 Higher": (p1 > p2).sum(),
        "Equal Values": (p1 == p2).sum()
    })

# Create a summary table
comparison_summary = pd.DataFrame(
    comparison_results
)

print("Comparison of 24 Overlapping Records:")
display(comparison_summary)

# Display all 24 participants for closer inspection
print("\nDetailed Step and Calorie Comparison:")

display(
    overlap_comparison[
        [
            "Id",
            "ActivityDate",
            "TotalSteps_p1",
            "TotalSteps_p2",
            "Calories_p1",
            "Calories_p2"
        ]
    ]
)
```

```python
# Step 2E.5: Resolve overlapping daily activity records

# Create copies and label their source periods
daily_p1 = daily_check_1.copy()
daily_p2 = daily_check_2.copy()

daily_p1["SourcePeriod"] = "Period 1"
daily_p2["SourcePeriod"] = "Period 2"

# Combine both datasets
daily_combined = pd.concat(
    [daily_p1, daily_p2],
    ignore_index=True
)

# Record the number of observations before cleaning
rows_before = len(daily_combined)

# Retain Period 2 for overlapping participant-date records
# Period 2 was appended last, so keep='last' gives it priority
daily_combined = daily_combined.drop_duplicates(
    subset=["Id", "ActivityDate"],
    keep="last"
).reset_index(drop=True)

# Record the number of observations after cleaning
rows_after = len(daily_combined)

# Validate the results
remaining_duplicates = daily_combined.duplicated(
    subset=["Id", "ActivityDate"]
).sum()

print("Records before cleaning:", rows_before)
print("Records after cleaning:", rows_after)
print("Overlapping records resolved:", rows_before - rows_after)
print("Remaining participant-date duplicates:", remaining_duplicates)

# Confirm the source retained for overlapping records
retained = daily_combined.merge(
    overlapping_records,
    on=["Id", "ActivityDate"],
    how="inner"
)

print("\nSource of retained overlapping records:")
print(retained["SourcePeriod"].value_counts())

# Ensure no participant-date duplicates remain
assert remaining_duplicates == 0
```

**Overlapping Records:** The two daily activity datasets contained 24 overlapping participant-date observations on April 12, 2016. All 24 Period 2 records reported higher total steps, distance, and calories than their Period 1 counterparts. This pattern suggests that Period 1 contains incomplete daily measurements for the overlapping date. Therefore, Period 2 was prioritized when resolving overlapping records. The original datasets were preserved, and the decision was documented for transparency.

```python
# Step 2F.1: Inspect invalid and unexpected values

# Identify activity measurement columns
# Exclude participant IDs, dates, and source labels
measurement_columns = daily_combined.select_dtypes(
    include="number"
).columns.drop("Id")

# Create a numeric quality summary
numeric_summary = pd.DataFrame({
    "Minimum": daily_combined[measurement_columns].min(),
    "Maximum": daily_combined[measurement_columns].max(),
    "Zero Values": (
        daily_combined[measurement_columns] == 0
    ).sum(),
    "Negative Values": (
        daily_combined[measurement_columns] < 0
    ).sum()
})

# Display results
print("Daily Activity Numeric Quality Summary:")
display(numeric_summary)

# Check whether recorded activity minutes exceed 24 hours
minute_columns = [
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes",
    "SedentaryMinutes"
]

# Calculate total recorded minutes per participant-day
total_minutes = daily_combined[minute_columns].sum(axis=1)

print("\nActivity Minute Validation:")
print("Days exceeding 1440 minutes:",
      (total_minutes > 1440).sum())

print("Days with zero recorded minutes:",
      (total_minutes == 0).sum())

# Check zero steps and calories
print("\nAdditional Checks:")
print("Days with zero steps:",
      (daily_combined["TotalSteps"] == 0).sum())

print("Days with zero calories:",
      (daily_combined["Calories"] == 0).sum())
```

```python

# Step 2F.2: Investigate zero-calorie observations

# Create a copy for further investigation
daily_inspection = daily_combined.copy()

# Calculate total recorded activity minutes
daily_inspection["TotalRecordedMinutes"] = (
    daily_inspection[
        [
            "VeryActiveMinutes",
            "FairlyActiveMinutes",
            "LightlyActiveMinutes",
            "SedentaryMinutes"
        ]
    ].sum(axis=1)
)

# Identify records with zero calories
zero_calories = daily_inspection[
    daily_inspection["Calories"] == 0
]

# Display relevant measurements
print("Records with Zero Calories:")

display(
    zero_calories[
        [
            "Id",
            "ActivityDate",
            "TotalSteps",
            "TotalDistance",
            "VeryActiveMinutes",
            "FairlyActiveMinutes",
            "LightlyActiveMinutes",
            "SedentaryMinutes",
            "TotalRecordedMinutes",
            "Calories",
            "SourcePeriod"
        ]
    ]
)

# Check whether these records also have zero steps
print("\nZero-calorie records with zero steps:",
      (zero_calories["TotalSteps"] == 0).sum())

print("Zero-calorie records with positive steps:",
      (zero_calories["TotalSteps"] > 0).sum())

# Check days with zero steps but positive calories
print("\nZero-step records with positive calories:",
      (
          (daily_inspection["TotalSteps"] == 0) &
          (daily_inspection["Calories"] > 0)
      ).sum())
```

```python
# Step 2F.3: Investigate zero-step records
# with positive calories

# Identify zero-step records with positive calories
zero_step_positive = daily_inspection[
    (daily_inspection["TotalSteps"] == 0) &
    (daily_inspection["Calories"] > 0)
].copy()

# Display total observations
print("Zero-step records with positive calories:",
      len(zero_step_positive))

# Summarize calorie measurements
print("\nCalorie Summary:")
display(
    zero_step_positive["Calories"].describe().round(2)
)

# Examine recorded activity minutes
print("\nRecorded Activity Minutes:")
display(
    zero_step_positive["TotalRecordedMinutes"]
    .describe().round(2)
)

# Check how many records contain 1440 sedentary minutes
print("\nRecords with 1440 sedentary minutes:",
      (zero_step_positive["SedentaryMinutes"] == 1440).sum())

# Count unique participants
print("Unique participants:",
      zero_step_positive["Id"].nunique())

# Compare the two collection periods
print("\nRecords by Source Period:")
display(
    zero_step_positive["SourcePeriod"].value_counts()
)

# Preview selected records
print("\nPreview of Zero-Step Records:")
display(
    zero_step_positive[
        [
            "Id",
            "ActivityDate",
            "TotalSteps",
            "SedentaryMinutes",
            "TotalRecordedMinutes",
            "Calories",
            "SourcePeriod"
        ]
    ].head(15)
)
```

```python
# Step 2F.4: Investigate unusual zero-step records

# Identify records with fewer than 1440 sedentary minutes
unusual_zero_steps = zero_step_positive[
    zero_step_positive["SedentaryMinutes"] < 1440
].copy()

# Display number of unusual records
print("Zero-step records with fewer than 1440 sedentary minutes:",
      len(unusual_zero_steps))

# Display the relevant measurements
display(
    unusual_zero_steps[
        [
            "Id",
            "ActivityDate",
            "TotalSteps",
            "TotalDistance",
            "VeryActiveMinutes",
            "FairlyActiveMinutes",
            "LightlyActiveMinutes",
            "SedentaryMinutes",
            "TotalRecordedMinutes",
            "Calories",
            "SourcePeriod"
        ]
    ]
)

# Identify zero-step records with positive active minutes
active_columns = [
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes"
]

active_with_zero_steps = (
    zero_step_positive[active_columns].sum(axis=1) > 0
)

print("\nZero-step records with positive active minutes:",
      active_with_zero_steps.sum())

# Count zero-step records with fewer than 1440 total minutes
print("Zero-step records with fewer than 1440 recorded minutes:",
      (zero_step_positive["TotalRecordedMinutes"] < 1440).sum())

# Identify unusually low calorie values
print("Zero-step records with fewer than 500 calories:",
      (zero_step_positive["Calories"] < 500).sum())
```

```python
# Step 2F.5: Create data quality flags

# Preserve all consolidated records
daily_final = daily_inspection.copy()

# Flag observations with zero calories
daily_final["FlagZeroCalories"] = (
    daily_final["Calories"] == 0
)

# Flag observations with zero steps
daily_final["FlagZeroSteps"] = (
    daily_final["TotalSteps"] == 0
)

# Flag days with fewer than 1440 recorded minutes
daily_final["FlagIncompleteMinutes"] = (
    daily_final["TotalRecordedMinutes"] < 1440
)

# Flag records with zero steps but positive active minutes
active_columns = [
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes"
]

daily_final["FlagActivityWithoutSteps"] = (
    (daily_final["TotalSteps"] == 0) &
    (daily_final[active_columns].sum(axis=1) > 0)
)

# Create a separate dataset for calorie analysis
# Exclude zero-calorie observations only
daily_calorie_analysis = daily_final[
    ~daily_final["FlagZeroCalories"]
].copy()

# Display record counts
print("Consolidated records:", len(daily_final))

print("Records for primary calorie analysis:",
      len(daily_calorie_analysis))

print("\nQuality Flag Summary:")

flag_columns = [
    "FlagZeroCalories",
    "FlagZeroSteps",
    "FlagIncompleteMinutes",
    "FlagActivityWithoutSteps"
]

display(
    daily_final[flag_columns]
    .sum()
    .to_frame(name="Flagged Records")
)

# Validate that original records are preserved
assert len(daily_final) == 1373

# Validate the primary calorie-analysis subset
assert len(daily_calorie_analysis) == 1367
```

**Data Quality — Daily Activity Measurements**

The consolidated daily activity dataset contained 1,373 unique participant-date observations after resolving overlapping records between the two collection periods.

Data-quality inspections identified 133 observations with zero steps, including six observations with zero calories. The remaining 127 zero-step observations contained positive calorie estimates.

Further examination of these 127 observations identified 11 records with fewer than 1,440 sedentary minutes, eight with fewer than 1,440 total recorded activity-category minutes, four with positive active minutes despite zero steps, and three with calorie estimates below 500.

Quality flags were created to identify potentially incomplete or inconsistent observations without permanently removing them from the consolidated dataset. The six zero-calorie records were excluded from the primary calorie-analysis subset, while the original consolidated observations were retained for additional analysis and validation.

```python
# Step 3E.2: Identify gaps in hourly observations

# Create a copy and sort by participant and timestamp
hourly_gap_check = hourly_combined.sort_values(
    ["Id", "ActivityHour"]
).copy()

# Identify each participant's previous recorded hour
hourly_gap_check["PreviousHour"] = (
    hourly_gap_check.groupby("Id")["ActivityHour"].shift()
)

# Calculate the time difference between consecutive records
hourly_gap_check["HourDifference"] = (
    (
        hourly_gap_check["ActivityHour"] -
        hourly_gap_check["PreviousHour"]
    ).dt.total_seconds() / 3600
)

# Identify gaps greater than one hour
hourly_gaps = hourly_gap_check[
    hourly_gap_check["HourDifference"] > 1
].copy()

# Calculate missing hourly slots between observations
hourly_gaps["MissingHoursBetween"] = (
    hourly_gaps["HourDifference"] - 1
).astype(int)

# Display summary results
print("Hourly Observation Gap Summary:")

print("Total gaps detected:", len(hourly_gaps))

print("Missing hourly slots between recorded observations:",
      hourly_gaps["MissingHoursBetween"].sum())

print("Participants with gaps:",
      hourly_gaps["Id"].nunique())

print("Total unique participants:",
      hourly_combined["Id"].nunique())

# Display the ten largest gaps
print("\nTop 10 Largest Gaps:")

display(
    hourly_gaps[
        [
            "Id",
            "PreviousHour",
            "ActivityHour",
            "MissingHoursBetween"
        ]
    ]
    .sort_values(
        "MissingHoursBetween",
        ascending=False
    )
    .head(10)
)
```

```python
# Step 6E.1: Inspect minute sleep recording continuity

# Create a copy to preserve the consolidated dataset
sleep_continuity = sleep_combined.copy()

# Sort observations by participant, sleep log, and time
sleep_continuity = sleep_continuity.sort_values(
    ["Id", "logId", "date"]
).reset_index(drop=True)

# Calculate time differences within each sleep log
sleep_continuity["MinutesSincePrevious"] = (
    sleep_continuity.groupby(["Id", "logId"])["date"]
    .diff()
    .dt.total_seconds() / 60
)

# Identify gaps greater than one minute
sleep_continuity["FlagGap"] = (
    sleep_continuity["MinutesSincePrevious"] > 1
)

# Identify unexpected non-minute intervals
intervals = sleep_continuity[
    "MinutesSincePrevious"
].dropna()

non_minute_intervals = (
    intervals.mod(1).round(8) != 0
).sum()

# Count recording gaps
total_gaps = sleep_continuity["FlagGap"].sum()

# Count sleep logs containing at least one gap
logs_with_gaps = (
    sleep_continuity.groupby(["Id", "logId"])["FlagGap"]
    .any()
    .sum()
)

# Count all unique sleep logs
total_logs = (
    sleep_continuity.groupby(["Id", "logId"])
    .ngroups
)

# Display recording continuity summary
print("Sleep Recording Continuity Summary:")

print("Total observations:", len(sleep_continuity))
print("Total unique sleep logs:", total_logs)
print("Total recording gaps:", total_gaps)
print("Sleep logs with gaps:", logs_with_gaps)
print("Sleep logs without gaps:",
      total_logs - logs_with_gaps)

print("Non-minute recording intervals:",
      non_minute_intervals)

# Display records immediately following gaps
sleep_gap_details = sleep_continuity[
    sleep_continuity["FlagGap"]
].copy()

print("\nTop 10 Largest Recording Gaps:")

display(
    sleep_gap_details[
        [
            "Id",
            "logId",
            "date",
            "MinutesSincePrevious",
            "SourcePeriod"
        ]
    ]
    .sort_values(
        "MinutesSincePrevious",
        ascending=False
    )
    .head(10)
)

# Summarize recording continuity by sleep log
sleep_log_summary = (
    sleep_continuity.groupby(["Id", "logId"])
    .agg(
        StartTime=("date", "min"),
        EndTime=("date", "max"),
        RecordedMinutes=("date", "size"),
        RecordingGaps=("FlagGap", "sum")
    )
    .reset_index()
)

# Calculate elapsed time from first to last observation
# Add one minute to include the first recorded minute
sleep_log_summary["ElapsedMinutes"] = (
    (
        sleep_log_summary["EndTime"] -
        sleep_log_summary["StartTime"]
    ).dt.total_seconds() / 60
) + 1

# Display recording duration statistics
print("\nSleep Log Duration Summary:")

display(
    sleep_log_summary[
        [
            "RecordedMinutes",
            "ElapsedMinutes",
            "RecordingGaps"
        ]
    ]
    .describe()
    .round(2)
)

# Verify expected dataset structure
assert len(sleep_continuity) == 382780
assert total_logs == 1007

print("\nSleep continuity inspection completed!")
```

```python
# Step 6E.2: Investigate unusual sleep-log durations

# Create a copy of the sleep-log summary
sleep_duration_check = sleep_log_summary.copy()

# 1. Flag unusually short sleep logs
# Less than 120 minutes (2 hours)
sleep_duration_check["FlagShortLog"] = (
    sleep_duration_check["RecordedMinutes"] < 120
)

# 2. Flag unusually long sleep logs
# More than 840 minutes (14 hours)
sleep_duration_check["FlagLongLog"] = (
    sleep_duration_check["RecordedMinutes"] > 840
)

# 3. Count flagged sleep logs
short_logs = sleep_duration_check["FlagShortLog"].sum()
long_logs = sleep_duration_check["FlagLongLog"].sum()

# 4. Display duration quality summary
print("Sleep Log Duration Quality Summary:")

print("Total sleep logs:", len(sleep_duration_check))
print("Logs shorter than 120 minutes:", short_logs)
print("Logs longer than 840 minutes:", long_logs)

print("Logs between 120 and 840 minutes:",
      (
          (~sleep_duration_check["FlagShortLog"]) &
          (~sleep_duration_check["FlagLongLog"])
      ).sum())

# 5. Display the ten shortest sleep logs
print("\nTop 10 Shortest Sleep Logs:")

display(
    sleep_duration_check[
        [
            "Id",
            "logId",
            "StartTime",
            "EndTime",
            "RecordedMinutes",
            "RecordingGaps"
        ]
    ]
    .sort_values(
        "RecordedMinutes",
        ascending=True
    )
    .head(10)
)

# 6. Display the ten longest sleep logs
print("\nTop 10 Longest Sleep Logs:")

display(
    sleep_duration_check[
        [
            "Id",
            "logId",
            "StartTime",
            "EndTime",
            "RecordedMinutes",
            "RecordingGaps"
        ]
    ]
    .sort_values(
        "RecordedMinutes",
        ascending=False
    )
    .head(10)
)

# 7. Identify participants with unusual sleep logs
print("\nParticipants with Short Sleep Logs:")

print(
    sleep_duration_check.loc[
        sleep_duration_check["FlagShortLog"], "Id"
    ].nunique()
)

print("\nParticipants with Long Sleep Logs:")

print(
    sleep_duration_check.loc[
        sleep_duration_check["FlagLongLog"], "Id"
    ].nunique()
)

# 8. Validate dataset structure and duration flags
assert len(sleep_duration_check) == 1007
assert (sleep_duration_check["RecordingGaps"] == 0).all()

assert (
    sleep_duration_check["FlagShortLog"] ==
    (sleep_duration_check["RecordedMinutes"] < 120)
).all()

assert (
    sleep_duration_check["FlagLongLog"] ==
    (sleep_duration_check["RecordedMinutes"] > 840)
).all()

print("\nSleep duration inspection completed!")
```

```python
# Step 6E.3: Examine sleep-state composition by sleep log

# 1. Count observations in each sleep state for every participant and sleep log
sleep_state_counts = (
    sleep_continuity.groupby(
        ["Id", "logId", "value"]
    )
    .size()
    .unstack(fill_value=0)
    .reindex(columns=[1, 2, 3], fill_value=0)
    .rename(columns={
        1: "State1Minutes",
        2: "State2Minutes",
        3: "State3Minutes"
    })
    .reset_index()
)

# 2. Combine sleep-state counts with log durations
sleep_state_summary = sleep_duration_check.merge(
    sleep_state_counts,
    on=["Id", "logId"],
    how="left",
    validate="one_to_one"
)

# 3. Calculate total state-classified minutes
sleep_state_summary["TotalStateMinutes"] = (
    sleep_state_summary[
        [
            "State1Minutes",
            "State2Minutes",
            "State3Minutes"
        ]
    ].sum(axis=1)
)

# 4. Calculate percentage of time in each state
for state in [1, 2, 3]:
    sleep_state_summary[f"State{state}Pct"] = (
        sleep_state_summary[f"State{state}Minutes"]
        / sleep_state_summary["RecordedMinutes"]
        * 100
    )

# 5. Display overall sleep-state statistics
print("Sleep-State Composition Summary:")

display(
    sleep_state_summary[
        [
            "State1Minutes",
            "State2Minutes",
            "State3Minutes",
            "State1Pct",
            "State2Pct",
            "State3Pct"
        ]
    ]
    .describe()
    .round(2)
)

# 6. Examine the nine unusually long sleep logs
print("\nSleep-State Composition of Long Logs:")

display(
    sleep_state_summary.loc[
        sleep_state_summary["FlagLongLog"],
        [
            "Id",
            "logId",
            "RecordedMinutes",
            "State1Minutes",
            "State2Minutes",
            "State3Minutes",
            "State1Pct",
            "State2Pct",
            "State3Pct"
        ]
    ]
    .sort_values(
        "RecordedMinutes",
        ascending=False
    )
    .round(2)
)

# 7. Examine the ten shortest sleep logs
print("\nSleep-State Composition of Shortest Logs:")

display(
    sleep_state_summary[
        [
            "Id",
            "logId",
            "RecordedMinutes",
            "State1Minutes",
            "State2Minutes",
            "State3Minutes",
            "State1Pct"
        ]
    ]
    .sort_values("RecordedMinutes")
    .head(10)
    .round(2)
)

# 8. Check for logs with no State 1 minutes
print("\nLogs with Zero State 1 Minutes:",
      (sleep_state_summary["State1Minutes"] == 0).sum())

# 9. Count logs at the observed maximum duration
print("Logs with Exactly 961 Minutes:",
      (sleep_state_summary["RecordedMinutes"] == 961).sum())

# 10. Validate state calculations
assert len(sleep_state_summary) == 1007

assert set(sleep_continuity["value"].unique()).issubset(
    {1, 2, 3}
)

assert (
    sleep_state_summary["TotalStateMinutes"] ==
    sleep_state_summary["RecordedMinutes"]
).all()

assert (
    sleep_state_summary[
        ["State1Minutes", "State2Minutes", "State3Minutes"]
    ].sum().sum() == 382780
)

print("\nSleep-state composition validation passed!")
```

```python
# Step 6F.1: Create final minute sleep quality flags

# 1. Create a copy of the sleep-log summary
sleep_logs_final = sleep_state_summary.copy()

# 2. Flag sleep logs containing exactly 961 minutes
# This is an exploratory quality flag, not proof of an error
sleep_logs_final["FlagDuration961"] = (
    sleep_logs_final["RecordedMinutes"] == 961
)

# 3. Define the episode-level quality flags
quality_flags = [
    "FlagShortLog",
    "FlagLongLog",
    "FlagDuration961"
]

# 4. Summarize flagged sleep logs
print("Sleep Log Quality Flag Summary:")

display(
    sleep_logs_final[quality_flags]
    .sum()
    .to_frame(name="Flagged Sleep Logs")
)

# 5. Create a copy of the cleaned minute-level dataset
sleep_final = sleep_continuity.copy()

# 6. Attach sleep-log quality flags to each minute
# Each minute inherits the flags of its sleep log
sleep_final = sleep_final.merge(
    sleep_logs_final[
        ["Id", "logId"] + quality_flags
    ],
    on=["Id", "logId"],
    how="left",
    validate="many_to_one"
)

# 7. Display final dataset dimensions
print("\nFinal Minute Sleep Dataset:")
print("Rows:", len(sleep_final))
print("Columns:", sleep_final.shape[1])

print("\nFinal Sleep Log Summary:")
print("Rows:", len(sleep_logs_final))
print("Columns:", sleep_logs_final.shape[1])

# 8. Verify all original observations were preserved
assert len(sleep_final) == 382780
assert len(sleep_logs_final) == 1007

# 9. Verify quality flags contain no missing values
assert sleep_final[quality_flags].notna().all().all()

# 10. Verify expected quality flag counts
assert sleep_logs_final["FlagShortLog"].sum() == 123
assert sleep_logs_final["FlagLongLog"].sum() == 9
assert sleep_logs_final["FlagDuration961"].sum() == 8

# 11. Verify the sleep-log summary accounts for every minute
assert (
    sleep_logs_final["RecordedMinutes"].sum()
    == len(sleep_final)
)

# 12. Verify no participant-minute duplicates remain
assert not sleep_final.duplicated(
    subset=["Id", "date"]
).any()

# 13. Verify no sleep logs have recording gaps
assert (sleep_logs_final["RecordingGaps"] == 0).all()

# 14. Display sample observations from both datasets
print("\nMinute-Level Dataset Preview:")
display(sleep_final.head())

print("\nSleep-Log Summary Preview:")
display(sleep_logs_final.head())

print("\nSleep quality flag preparation successful!")
```

```python
# Step 6F.2: Finalize minute sleep dataset structures

# ==========================================
# PART 1: Prepare minute-level observations
# ==========================================

# Select columns needed for future analysis
# Temporary diagnostic columns remain available in sleep_continuity and sleep_final
minute_columns = [
    "Id",
    "date",
    "value",
    "logId",
    "SourcePeriod",
    "FlagShortLog",
    "FlagLongLog",
    "FlagDuration961"
]

# Create the organized minute-level dataset
sleep_minutes_ready = (
    sleep_final[minute_columns]
    .sort_values(["Id", "logId", "date"])
    .reset_index(drop=True)
    .copy()
)

# ==========================================
# PART 2: Prepare sleep-log summary
# ==========================================

# Create a copy of the sleep-log summary
sleep_logs_ready = sleep_logs_final.copy()

# Extract the calendar date when each log ends
# Overnight logs are assigned their ending date
sleep_logs_ready["SleepEndDate"] = (
    sleep_logs_ready["EndTime"].dt.date
)

# Convert recorded minutes to hours
sleep_logs_ready["RecordedHours"] = (
    sleep_logs_ready["RecordedMinutes"] / 60
).round(2)

# Organize columns into a logical order
log_columns = [
    "Id",
    "logId",
    "StartTime",
    "EndTime",
    "SleepEndDate",
    "RecordedMinutes",
    "RecordedHours",
    "ElapsedMinutes",
    "RecordingGaps",
    "State1Minutes",
    "State2Minutes",
    "State3Minutes",
    "TotalStateMinutes",
    "State1Pct",
    "State2Pct",
    "State3Pct",
    "FlagShortLog",
    "FlagLongLog",
    "FlagDuration961"
]

# Select and sort the final sleep-log columns
sleep_logs_ready = (
    sleep_logs_ready[log_columns]
    .sort_values(["Id", "StartTime", "logId"])
    .reset_index(drop=True)
)

# ==========================================
# PART 3: Verify both dataset structures
# ==========================================

# Verify all observations were preserved
assert sleep_minutes_ready.shape == (382780, 8)
assert sleep_logs_ready.shape == (1007, 19)

# Verify participant-minute uniqueness
assert not sleep_minutes_ready.duplicated(
    subset=["Id", "date"]
).any()

# Verify sleep-log uniqueness
assert not sleep_logs_ready.duplicated(
    subset=["Id", "logId"]
).any()

# Verify no recorded minutes were lost
assert (
    sleep_logs_ready["RecordedMinutes"].sum()
    == len(sleep_minutes_ready)
)

# Verify quality flags remain unchanged
assert sleep_logs_ready["FlagShortLog"].sum() == 123
assert sleep_logs_ready["FlagLongLog"].sum() == 9
assert sleep_logs_ready["FlagDuration961"].sum() == 8

# ==========================================
# PART 4: Display the final structures
# ==========================================

print("Final Minute-Level Dataset:")
print("Shape:", sleep_minutes_ready.shape)
print("Columns:", sleep_minutes_ready.columns.tolist())

print("\nFinal Sleep-Log Summary:")
print("Shape:", sleep_logs_ready.shape)
print("Columns:", sleep_logs_ready.columns.tolist())

print("\nMinute-Level Preview:")
display(sleep_minutes_ready.head())

print("\nSleep-Log Summary Preview:")
display(sleep_logs_ready.head())

print("\nSleep dataset structure preparation successful!")
```

```python

# Step 7B.1: Inspect missing and invalid heart rate values

# ==========================================
# 1. Inspect each collection period
# ==========================================

for label, df in [
    ("Period 1", heart_1),
    ("Period 2", heart_2)
]:

    print(f"\n===== {label} =====")

    # Display total records
    print("Total observations:", len(df))

# ======================================
# 2. Check missing values
# ======================================

    print("\nMissing Values:")

    missing_counts = df.isna().sum()

    missing_percent = (
        missing_counts / len(df) * 100
    )

    # Create missing value summary table
    missing_summary = pd.DataFrame({
        "Missing Count": missing_counts,
        "Missing (%)": missing_percent
    })

    display(missing_summary.round(2))

    print(
        "Total missing values:",
        missing_counts.sum()
    )

# ======================================
# 3. Check invalid timestamps
# ======================================

    invalid_timestamps = df["Time"].isna().sum()

    print("\nTimestamp Validation:")

    print(
        "Invalid or missing timestamps:",
        invalid_timestamps
    )

    print(
        "Timestamp data type:",
        df["Time"].dtype
    )

# ======================================
# 4. Check heart rate measurements
# ======================================

    print("\nHeart Rate Value Checks:")

    # Missing heart rate observations
    missing_hr = df["Value"].isna().sum()

    # Nonpositive values are invalid as
    # recorded heart rates
    zero_hr = (df["Value"] == 0).sum()
    negative_hr = (df["Value"] < 0).sum()

    print("Missing heart rate values:", missing_hr)
    print("Zero heart rate values:", zero_hr)
    print("Negative heart rate values:", negative_hr)

# ======================================
# 5. Preview problematic records
# ======================================

    problem_mask = (
        df["Id"].isna()
        | df["Time"].isna()
        | df["Value"].isna()
        | (df["Value"] <= 0).fillna(False)
    )

    problem_records = df.loc[problem_mask]

    print("\nRecords requiring investigation:",
          len(problem_records))

    display(problem_records.head(10))

print("\nHeart rate missing-value inspection completed!")
```

```python
# Step 7B.2: Inspect heart rate recording intervals

# ==========================================
# 1. Inspect each collection period
# ==========================================

for label, df in [
    ("Period 1", heart_1),
    ("Period 2", heart_2)
]:

    print(f"\n===== {label} =====")

    # Create a temporary copy with relevant columns
    intervals_df = df[
        ["Id", "Time"]
    ].copy()

    # Sort observations by participant and time
    intervals_df = intervals_df.sort_values(
        ["Id", "Time"]
    )

# ======================================
# 2. Calculate time between observations
# ======================================

    # Calculate intervals separately for each participant
    intervals_df["SecondsSincePrevious"] = (
        intervals_df
        .groupby("Id", observed=True)["Time"]
        .diff()
        .dt.total_seconds()
    )

    # Exclude the first observation per participant,
    # which has no previous timestamp
    intervals = (
        intervals_df["SecondsSincePrevious"]
        .dropna()
    )

# ======================================
# 3. Display interval statistics
# ======================================

    print("\nRecording Interval Statistics:")

    print("Total intervals:", len(intervals))

    display(intervals.describe().round(2))

# ======================================
# 4. Identify recording frequencies
# ======================================

    print("\nTop 10 Most Common Intervals (Seconds):")

    interval_frequency = (
        intervals.value_counts()
        .head(10)
        .to_frame(name="Observations")
    )

    display(interval_frequency)

# ======================================
# 5. Check irregular intervals
# ======================================

    print("\nRecording Interval Checks:")

    print(
        "Exactly 5 seconds:",
        (intervals == 5).sum()
    )

    print(
        "Less than 5 seconds (positive):",
        ((intervals > 0) & (intervals < 5)).sum()
    )

    print(
        "Zero-second intervals:",
        (intervals == 0).sum()
    )

    print(
        "Intervals greater than 5 seconds:",
        (intervals > 5).sum()
    )

    print(
        "Intervals greater than 60 seconds:",
        (intervals > 60).sum()
    )

    print(
        "Intervals greater than 1 hour:",
        (intervals > 3600).sum()
    )

# ======================================
# 6. Calculate recording consistency
# ======================================

    five_second_pct = (
        (intervals == 5).mean() * 100
    )

    print(
        "\nPercentage at 5-second intervals:",
        round(five_second_pct, 2),
        "%"
    )

# ======================================
# 7. Identify largest recording gap
# ======================================

    print(
        "\nLargest recording gap (hours):",
        round(intervals.max() / 3600, 2)
    )

print("\nHeart rate interval inspection completed!")
```

```python

# Step 7C.1: Identify duplicate heart rate records

# ==========================================
# 1. Inspect each collection period
# ==========================================

for label, df in [
    ("Period 1", heart_1),
    ("Period 2", heart_2)
]:

    print(f"\n===== {label} =====")

# ======================================
# 2. Check exact duplicate records
# ======================================

    exact_duplicates = df.duplicated().sum()

    print("\nExact Duplicate Records:")
    print("Total:", exact_duplicates)

# ======================================
# 3. Check duplicate participant-times
# ======================================

    key_columns = ["Id", "Time"]

    duplicate_keys = df.duplicated(
        subset=key_columns
    ).sum()

    print("\nDuplicate Participant-Timestamp Records:")
    print("Total:", duplicate_keys)

# ======================================
# 4. Display duplicate observations
# ======================================

    duplicate_details = df.loc[
        df.duplicated(
            subset=key_columns,
            keep=False
        ),
        ["Id", "Time", "Value"]
    ].sort_values(key_columns)

    print("\nPreview of Duplicate Records:")
    display(duplicate_details.head(20))

# ======================================
# 5. Check conflicting measurements
# ======================================

    if duplicate_details.empty:
        conflicting_keys = 0

    else:
        # Count timestamp groups with
        # more than one heart rate value
        conflicting_keys = (
            duplicate_details
            .groupby(
                key_columns,
                observed=True
            )["Value"]
            .nunique()
            .gt(1)
            .sum()
        )

    print("\nConflicting Participant-Timestamps:")
    print("Total:", conflicting_keys)

# ======================================
# 6. Display duplicate summary
# ======================================

    print("\nDuplicate Summary:")

    print("Total observations:", len(df))
    print("Exact duplicates:", exact_duplicates)
    print("Duplicate participant-times:", duplicate_keys)
    print("Conflicting timestamp groups:", conflicting_keys)

print("\nHeart rate duplicate inspection completed!")
```

```python

# Step 7D.1: Identify overlapping heart rate records

# ==========================================
# 1. Define the overlapping calendar date
# ==========================================

overlap_date = pd.Timestamp("2016-04-12")

# ==========================================
# 2. Extract records from the overlapping date
# ==========================================

heart_overlap_1 = heart_1.loc[
    heart_1["Time"].dt.normalize() == overlap_date,
    ["Id", "Time", "Value"]
].copy()

heart_overlap_2 = heart_2.loc[
    heart_2["Time"].dt.normalize() == overlap_date,
    ["Id", "Time", "Value"]
].copy()

# Standardize participant IDs for comparison
heart_overlap_1["Id"] = (
    heart_overlap_1["Id"].astype("string")
)

heart_overlap_2["Id"] = (
    heart_overlap_2["Id"].astype("string")
)

# ==========================================
# 3. Display records on April 12
# ==========================================

print("Heart Rate Overlap Inspection:")

print("\nRecords on April 12, 2016:")

print("Period 1:", len(heart_overlap_1))
print("Period 2:", len(heart_overlap_2))

print("\nUnique participants on April 12:")

print("Period 1:", heart_overlap_1["Id"].nunique())
print("Period 2:", heart_overlap_2["Id"].nunique())

# ==========================================
# 4. Match participant-timestamp observations
# ==========================================

heart_overlap = heart_overlap_1.merge(
    heart_overlap_2,
    on=["Id", "Time"],
    how="inner",
    suffixes=("_p1", "_p2"),
    validate="one_to_one"
)

print("\nTotal overlapping participant-timestamp records:")
print(len(heart_overlap))

# ==========================================
# 5. Compare heart rate measurements
# ==========================================

heart_overlap["HeartRateMatch"] = (
    heart_overlap["Value_p1"] ==
    heart_overlap["Value_p2"]
)

matching_records = (
    heart_overlap["HeartRateMatch"].sum()
)

different_records = (
    (~heart_overlap["HeartRateMatch"]).sum()
)

print("\nHeart Rate Measurement Comparison:")

print("Matching measurements:", matching_records)
print("Different measurements:", different_records)

# ==========================================
# 6. Display overlapping records
# ==========================================

print("\nPreview of Overlapping Records:")

display(heart_overlap.head(15))

# ==========================================
# 7. Identify conflicting measurements
# ==========================================

conflicting_records = heart_overlap.loc[
    ~heart_overlap["HeartRateMatch"]
]

print("\nPreview of Conflicting Records:")

display(conflicting_records.head(15))

# ==========================================
# 8. Summarize overlap
# ==========================================

print("\nOverlap Summary:")

print(
    "Total matching participant-timestamps:",
    len(heart_overlap)
)

print(
    "Identical heart rate measurements:",
    matching_records
)

print(
    "Conflicting heart rate measurements:",
    different_records
)

print("\nHeart rate overlap inspection completed!")
```

```python
# Step 7D.2: Consolidate heart rate datasets

# ==========================================
# 1. Prepare both collection periods
# ==========================================

# Create copies to preserve original datasets
heart_period1 = heart_1.copy()
heart_period2 = heart_2.copy()

# Label each record by its collection period
heart_period1["SourcePeriod"] = "Period 1"
heart_period2["SourcePeriod"] = "Period 2"

# ==========================================
# 2. Combine the datasets
# ==========================================

# Append Period 2 after Period 1
heart_combined = pd.concat(
    [heart_period1, heart_period2],
    ignore_index=True
)

# Record the original number of observations
rows_before = len(heart_combined)

# ==========================================
# 3. Resolve overlapping records
# ==========================================

# Keep Period 2 when a participant and timestamp
# appear in both collection periods
heart_combined = heart_combined.drop_duplicates(
    subset=["Id", "Time"],
    keep="last"
)

# Sort observations chronologically by participant
heart_combined = heart_combined.sort_values(
    ["Id", "Time"]
).reset_index(drop=True)

# ==========================================
# 4. Calculate consolidation results
# ==========================================

rows_after = len(heart_combined)

records_removed = rows_before - rows_after

remaining_duplicates = heart_combined.duplicated(
    subset=["Id", "Time"]
).sum()

print("Heart Rate Consolidation Summary:")

print("Records before consolidation:", rows_before)
print("Records after consolidation:", rows_after)
print("Overlapping records removed:", records_removed)

print(
    "Remaining participant-timestamp duplicates:",
    remaining_duplicates
)

# ==========================================
# 5. Verify retained source periods
# ==========================================

print("\nRecords by Source Period:")

print(heart_combined["SourcePeriod"].value_counts())

# Verify all April 12 observations
# were retained from Period 2
april12_records = heart_combined.loc[
    heart_combined["Time"].dt.normalize()
    == pd.Timestamp("2016-04-12")
]

print("\nApril 12 Records by Source Period:")

print(april12_records["SourcePeriod"].value_counts())

# ==========================================
# 6. Validate consolidation
# ==========================================

assert rows_before == 3638339
assert rows_after == 3614915
assert records_removed == 23424
assert remaining_duplicates == 0

assert (
    april12_records["SourcePeriod"] == "Period 2"
).all()

print("\nHeart rate consolidation successful!")

# ==========================================
# 7. Preview consolidated dataset
# ==========================================

print("\nConsolidated Dataset Preview:")

display(heart_combined.head(10))

print("\nConsolidated Dataset Shape:")
print(heart_combined.shape)
```

**Overlapping Records — Heart Rate**

The two heart rate datasets initially contained 3,638,339 observations across both collection periods. Data-quality inspections identified 23,424 overlapping participant-timestamp observations on April 12, 2016. All overlapping observations contained identical heart rate measurements.

To prevent duplicate counting, Period 2 was prioritized when consolidating the datasets. One copy of each overlapping participant-timestamp observation was retained, while all nonoverlapping observations were preserved.

This procedure reduced the combined dataset to 3,614,915 unique participant-timestamp observations. The original source datasets remained unchanged.

```python
# Step 7E.2: Investigate extreme heart rate patterns

# ==========================================
# 1. Identify extreme heart rate observations
# ==========================================

extreme_heart = heart_combined.loc[
    (heart_combined["Value"] < 40) |
    (heart_combined["Value"] > 180)
].copy()

print("Extreme Heart Rate Investigation:")

print("\nTotal extreme observations:", len(extreme_heart))

print(
    "Unique participants:",
    extreme_heart["Id"].nunique()
)

# ==========================================
# 2. Summarize extreme values by participant
# ==========================================

extreme_heart["Below40"] = (
    extreme_heart["Value"] < 40
)

extreme_heart["Above180"] = (
    extreme_heart["Value"] > 180
)

participant_summary = (
    extreme_heart
    .groupby("Id", observed=True)
    .agg(
        TotalExtreme=("Value", "size"),
        Below40=("Below40", "sum"),
        Above180=("Above180", "sum"),
        MinimumBPM=("Value", "min"),
        MaximumBPM=("Value", "max")
    )
    .sort_values(
        "TotalExtreme",
        ascending=False
    )
)

print("\nExtreme Measurements by Participant:")
display(participant_summary)

# ==========================================
# 3. Investigate surrounding measurements
# ==========================================

# Identify examples of the lowest and highest values
lowest_record = heart_combined.loc[
    heart_combined["Value"].idxmin()
]

highest_record = heart_combined.loc[
    heart_combined["Value"].idxmax()
]

# Display readings within three minutes
# before and after each extreme observation
def inspect_heart_context(record, label):

    participant = record["Id"]
    timestamp = record["Time"]

    start = timestamp - pd.Timedelta(minutes=3)
    end = timestamp + pd.Timedelta(minutes=3)

    context = heart_combined.loc[
        (heart_combined["Id"] == participant) &
        (heart_combined["Time"].between(start, end))
    ].copy()

    context = context.sort_values("Time")

    print(f"\n{label}:")

    print("Participant:", participant)
    print("Extreme timestamp:", timestamp)
    print("Extreme heart rate:", record["Value"])

    print("\nSurrounding Measurements:")
    display(
        context[
            ["Id", "Time", "Value", "SourcePeriod"]
        ]
    )

# ==========================================
# 4. Examine lowest recorded heart rate
# ==========================================

inspect_heart_context(
    lowest_record,
    "Lowest Heart Rate Context"
)

# ==========================================
# 5. Examine highest recorded heart rate
# ==========================================

inspect_heart_context(
    highest_record,
    "Highest Heart Rate Context"
)

# ==========================================
# 6. Summarize extreme readings by date
# ==========================================

extreme_heart["ActivityDate"] = (
    extreme_heart["Time"].dt.normalize()
)

print("\nTop 15 Dates with Extreme Measurements:")

date_summary = (
    extreme_heart
    .groupby("ActivityDate")
    .agg(
        ExtremeRecords=("Value", "size"),
        Participants=("Id", "nunique"),
        MinimumBPM=("Value", "min"),
        MaximumBPM=("Value", "max")
    )
    .sort_values(
        "ExtremeRecords",
        ascending=False
    )
)

display(date_summary.head(15))

print("\nExtreme heart rate investigation completed!")
```

```python
# Step 7E.3: Create heart rate quality flags

# ==========================================
# 1. Create a working copy of the dataset
# ==========================================

heart_final = heart_combined.copy()

# Sort observations by participant and time
heart_final = (
    heart_final
    .sort_values(["Id", "Time"])
    .reset_index(drop=True)
)

# ==========================================
# 2. Create heart rate threshold flags
# ==========================================

# Very low heart rate
heart_final["FlagBelow40"] = (
    heart_final["Value"] < 40
)

# Low heart rate screening category
heart_final["Flag40to49"] = (
    heart_final["Value"].between(40, 49)
)

# High heart rate
heart_final["FlagAbove180"] = (
    heart_final["Value"] > 180
)

# Very high heart rate
heart_final["FlagAbove200"] = (
    heart_final["Value"] > 200
)

# ==========================================
# 3. Calculate time between consecutive readings
# ==========================================

heart_final["SecondsSincePrevious"] = (
    heart_final
    .groupby("Id", observed=True)["Time"]
    .diff()
    .dt.total_seconds()
)

# ==========================================
# 4. Calculate change in heart rate
# ==========================================

heart_final["BPMChange"] = (
    heart_final
    .groupby("Id", observed=True)["Value"]
    .diff()
)

# ==========================================
# 5. Flag unusually rapid changes
# ==========================================

# Screening rule:
# Change greater than 40 BPM within 30 seconds
# compared with the immediately previous reading

heart_final["FlagRapidChange"] = (
    heart_final["SecondsSincePrevious"].between(
        0, 30,
        inclusive="right"
    )
    &
    (heart_final["BPMChange"].abs() > 40)
)

# ==========================================
# 6. Summarize quality flags
# ==========================================

flag_columns = [
    "FlagBelow40",
    "Flag40to49",
    "FlagAbove180",
    "FlagAbove200",
    "FlagRapidChange"
]

print("Heart Rate Quality Flag Summary:")

display(
    heart_final[flag_columns]
    .sum()
    .to_frame(name="Flagged Records")
)

# ==========================================
# 7. Summarize flags by source period
# ==========================================

print("\nQuality Flags by Source Period:")

display(
    heart_final
    .groupby("SourcePeriod", observed=True)[flag_columns]
    .sum()
)

# ==========================================
# 8. Inspect the largest rapid changes
# ==========================================

print("\nTop 20 Largest Flagged Rapid Changes:")

rapid_changes = heart_final.loc[
    heart_final["FlagRapidChange"]
].copy()

rapid_changes["AbsoluteBPMChange"] = (
    rapid_changes["BPMChange"].abs()
)

display(
    rapid_changes
    .sort_values(
        "AbsoluteBPMChange",
        ascending=False
    )
    [
        [
            "Id",
            "Time",
            "Value",
            "BPMChange",
            "SecondsSincePrevious",
            "SourcePeriod"
        ]
    ]
    .head(20)
)

# ==========================================
# 9. Validate original records and flags
# ==========================================

# Ensure no observations were removed
assert len(heart_final) == 3614915

# Ensure participant-timestamp uniqueness
assert not heart_final.duplicated(
    subset=["Id", "Time"]
).any()

# Verify the original extreme measurement counts
assert heart_final["FlagBelow40"].sum() == 37
assert heart_final["Flag40to49"].sum() == 52977
assert heart_final["FlagAbove180"].sum() == 250
assert heart_final["FlagAbove200"].sum() == 13

# Verify that all above-200 readings are also above 180
assert (
    ~heart_final["FlagAbove200"]
    | heart_final["FlagAbove180"]
).all()

print("\nOriginal records preserved:", len(heart_final))
print("Final dataset shape:", heart_final.shape)

print("\nHeart rate quality flag preparation completed!")
```
